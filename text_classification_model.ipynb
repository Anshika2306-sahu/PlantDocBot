```python

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report
import joblib
```

```python

df = pd.read_csv("dataset.csv")
print("Dataset loaded:", df.shape)
df.head()
```

```python
text_col = "captions"
label_col = "caption"

df = df.dropna(subset=[text_col, label_col])

print("Unique labels:", df[label_col].nunique())
print("Sample data:")
print(df.head())
```

```python

X_train, X_test, y_train, y_test = train_test_split(
    df[text_col], df[label_col], test_size=0.2, random_state=42, stratify=df[label_col]
)
```

```python

vectorizer = TfidfVectorizer(max_features=5000, stop_words='english')
X_train_tfidf = vectorizer.fit_transform(X_train)
X_test_tfidf = vectorizer.transform(X_test)

print("Vectorization complete.")
```

```python

model = LogisticRegression(max_iter=1000)
model.fit(X_train_tfidf, y_train)
print("Model training complete.")
```

```python

y_pred = model.predict(X_test_tfidf)
print("Accuracy:", accuracy_score(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred))
```

```python
import os

model_dir = "model"
os.makedirs(model_dir, exist_ok=True)
joblib.dump(model, "model/model.pkl")
joblib.dump(vectorizer, "model/tfidf_vectorizer.pkl")
print("Model and vectorizer saved successfully.")
```

```python

sample_text = ["leaf with yellow spots", "healthy green leaf"]
sample_vec = vectorizer.transform(sample_text)
predictions = model.predict(sample_vec)
list(zip(sample_text, predictions))
```
